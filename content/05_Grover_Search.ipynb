# 05 — Grover Search

**Concept.** Grover finds a marked item in ~sqrt(N) steps — quadratic speedup.

```python
from lqe.core.statevector import StatevectorSim
import numpy as np

sim = StatevectorSim(2)
sim.apply('H', 0); sim.apply('H', 1)
for it in range(3):
    st = sim.state.copy(); st[3] *= -1; sim.state = st
    sim.apply('H', 0); sim.apply('H', 1)
    st = sim.state.copy(); mean = np.mean(st); sim.state = 2*mean - st
    sim.apply('H', 0); sim.apply('H', 1)
    print(f'iters={it}: P(marked)={sim.probabilities()[3]:.4f}')
```

**Takeaway.** One optimal Grover iteration concentrates amplitude on the marked state.
