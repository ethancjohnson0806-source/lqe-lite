This lesson is adapted for the pure-NumPy browser build. See the full LQE repository for unavailable modules.


```python
from lqe import VQE, tfi_hamiltonian
from lqe.circuit_tools.qasm_export import export_qasm
H = tfi_hamiltonian(4)
out = VQE(4, H, num_layers=2, max_iterations=60).solve()
print(f"VQE energy={out['energy']:.6f}")
print(export_qasm([('h', [q]) for q in range(4)], 4))
```
