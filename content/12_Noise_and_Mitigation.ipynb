# 12 — Noise and Mitigation

**Concept.** Depolarizing noise destroys purity; mitigation recovers signal.

```python
from lqe.core.statevector import StatevectorSim
from lqe.core.noise_models import depolarizing_channel
import numpy as np

sim = StatevectorSim(2)
sim.apply('H', 0); sim.apply('CNOT', 0, 1)
psi = sim.state.reshape(-1, 1)
rho = psi @ psi.conj().T
for p in [0.0, 0.1, 0.3]:
    rn = depolarizing_channel(rho, n_qubits=2, p=p)
    print(f'p={p:.2f}  purity={np.real(np.trace(rn@rn)):.4f}')
```

**Takeaway.** Even modest noise destroys purity; QEC and mitigation fight this.
