# 03 — Entanglement

**Concept.** Entangled qubits cannot be written as a product of individual states.

Why it matters: entanglement is a resource for teleportation and many algorithms.

**Math.**

Bell state: $$|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$$

```python
from lqe.core.statevector import StatevectorSim
import numpy as np

sim = StatevectorSim(2)
sim.apply("H", 0)
sim.apply("CNOT", 0, 1)
print("Bell |Phi+>:", np.round(sim.state, 4))
print("Probs:", np.round(sim.probabilities(), 4))
```
