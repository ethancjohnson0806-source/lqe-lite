# 01 — What Is A Qubit?

**Concept.** A classical bit is either 0 or 1. A qubit is a two-level quantum system that can be in a superposition of both.

Why it matters: every quantum algorithm starts here.

**Math.**

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle, \quad |\alpha|^2 + |\beta|^2 = 1$$

```python
from lqe.core.statevector import StatevectorSim
import numpy as np

sim = StatevectorSim(1)
print("Initial |0>:", sim.state)
sim.apply("X", 0)
print("After X  |1>:", sim.state)
sim = StatevectorSim(1)
sim.apply("H", 0)
print("After H  |+>:", sim.state)
print("Probabilities:", sim.probabilities())
```

**Exercise.** Apply H twice. What state do you get back?

**Takeaway.** A qubit is a normalized 2-vector; gates are unitary matrices.
