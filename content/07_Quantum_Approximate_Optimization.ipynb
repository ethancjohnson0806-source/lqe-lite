# 07 — Quantum Approximate Optimization (QAOA)

**Concept.** QAOA is a hybrid variational algorithm for combinatorial optimization. It alternates cost and mixer unitaries, with parameters optimized classically.

Why it matters: MaxCut and related problems are the standard demo of near-term quantum optimization.

**Math.**

Cost Hamiltonian for MaxCut encodes edge penalties.

```python
from lqe.solvers.qaoa import QAOA
import numpy as np

edges = [(0, 1), (1, 2), (2, 0)]
n = 3

for p in [1, 2]:
    qaoa = QAOA(n, edges, p=p, max_iterations=60)
    out = qaoa.solve(verbose=False)
    print(f"p={p}: energy={out['energy']:.6f}  exact={out['exact']:.6f}  error={out['error']:.6f}")
```

**Takeaway.** QAOA depth p controls approximation quality; even p=1 often captures useful structure on small graphs.
