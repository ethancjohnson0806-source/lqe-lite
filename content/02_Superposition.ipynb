# 02 — Superposition

**Concept.** Superposition lets a qubit hold many basis states at once. Interference is what makes quantum algorithms powerful.

Why it matters: without interference, superposition is just fancy randomness.

**Math.**

$$H|0\rangle = |+\rangle,\quad H^2 = I$$

```python
from lqe.core.statevector import StatevectorSim
import numpy as np

sim = StatevectorSim(1)
sim.apply("H", 0)
print("After one H:", np.round(sim.state, 4))
sim.apply("H", 0)
print("After two H:", np.round(sim.state, 4))

sim2 = StatevectorSim(2)
sim2.apply("H", 0)
sim2.apply("H", 1)
print("2-qubit uniform:", np.round(sim2.probabilities(), 4))
```

**Exercise.** Apply H, then Z, then H. What state do you get?

**Takeaway.** Superposition + interference is the engine of quantum advantage.
