# 08 — Tensor Networks (MPS)

**Concept.** MPS compress area-law states enabling simulation beyond full statevectors.

```python
from lqe.tensor_networks.mps_simulator import mps_ghz_state
from lqe.core.statevector import StatevectorSim
import numpy as np

n=4
mps = mps_ghz_state(n)
sim = StatevectorSim(n)
sim.apply('H', 0)
for q in range(n-1): sim.apply('CNOT', q, q+1)
print(f'MPS type: {type(mps).__name__}')
print(f'GHZ nonzero: {np.where(np.abs(sim.state)>0.1)[0]}')
```

**Takeaway.** Area-law states stay compressible with MPS.
