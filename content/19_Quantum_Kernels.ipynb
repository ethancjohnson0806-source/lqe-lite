# 19 — Quantum Kernels

This lesson is adapted for the pure-NumPy browser build. See the full LQE repository for unavailable modules.


```python
import numpy as np
from lqe import StatevectorSim
def fidelity(x, y):
    a, b = StatevectorSim(2), StatevectorSim(2)
    for q, angle in enumerate(x): a.apply("RY", float(angle), q)
    for q, angle in enumerate(y): b.apply("RY", float(angle), q)
    return float(abs(np.vdot(a.state, b.state)) ** 2)
X = np.array([[0.1, 0.2], [0.8, 0.7]])
print(np.array([[fidelity(x, y) for y in X] for x in X]).round(4))
```
