# Welcome to LQE Lite

**A quantum computing environment that runs in your browser.**

No installation. No server. No account. Just quantum code.

## First time?

1. Wait for the environment to load.
2. Open `01_What_Is_A_Qubit.ipynb` in the file browser.
3. Run a code cell and edit it freely.

## What works

- Pure-NumPy statevector simulation
- VQE and QAOA solvers
- MPS tensor-network demonstrations
- Stabilizer simulation
- Twenty adapted educational notebooks

## Honest limitations

- No Numba, SciPy, OpenFermion, PySCF, networkx, or compiled extensions.
- Some advanced lessons are simplified browser-safe demos.
- Memory and speed depend on your phone or browser.
- Work is saved in browser storage; there is no server account.

**Ready?** Open `01_What_Is_A_Qubit.ipynb` and click Run.
