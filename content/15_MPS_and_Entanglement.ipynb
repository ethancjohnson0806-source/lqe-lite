# 15 — MPS and Entanglement

**Concept.** Entanglement entropy diagnoses when MPS works.

```python
from lqe.core.statevector import StatevectorSim
import numpy as np

def bipartition_entropy(state, n, cut):
    psi = state.reshape(2**cut, 2**(n-cut))
    s = np.linalg.svd(psi, compute_uv=False)
    p = s**2; p = p[p > 1e-15]
    return float(-np.sum(p * np.log2(p)))

n=6
sim = StatevectorSim(n)
for q in range(n): sim.apply('H', q)
print(f'Product S={bipartition_entropy(sim.state, n, 3):.4f}')
sim2 = StatevectorSim(n)
sim2.apply('H', 0)
for q in range(n-1): sim2.apply('CNOT', q, q+1)
print(f'GHZ S={bipartition_entropy(sim2.state, n, 3):.4f}')
```

**Takeaway.** Area-law states stay MPS-compressible.
