# 06 — VQE

**Concept.** VQE finds ground-state energies with a parameterized trial state optimized classically.

```python
from lqe.solvers.vqe_ground import VQE, tfi_hamiltonian
import numpy as np

H = tfi_hamiltonian(4, J=1.0, h=0.5)
exact = float(np.min(np.linalg.eigvalsh(H)))
out = VQE(4, H, num_layers=2, max_iterations=60).solve(verbose=False)
print(f'Exact: {exact:.6f}')
print(f'VQE:   {out["energy"]:.6f}  status={out["status"]}')
```

**Takeaway.** VQE is the workhorse hybrid algorithm for near-term chemistry.
