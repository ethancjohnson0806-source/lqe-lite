# 16 — Quantum Volume

This lesson is adapted for the pure-NumPy browser build. See the full LQE repository for unavailable modules.


```python
import numpy as np
from lqe import StatevectorSim
sim = StatevectorSim(3)
for q in range(3):
    sim.apply("H", q)
print("Pure-NumPy quantum-volume-style smoke test")
print("Uniform probabilities:", np.round(sim.probabilities(), 4))
print("SciPy-based Haar random circuits are intentionally excluded from LQE Lite.")
```
