# 13 — Solver Orchestration

This lesson is adapted for the pure-NumPy browser build. See the full LQE repository for unavailable modules.


```python
import numpy as np
from lqe import auto_solve, tfi_hamiltonian
for n in (2, 3, 4):
    out = auto_solve(tfi_hamiltonian(n), n, max_iterations=40)
    print(f"{n} qubits: energy={out['energy']:.6f}, status={out['status']}")
```
