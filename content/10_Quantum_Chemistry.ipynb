# 10 — Quantum Chemistry

This lesson is adapted for the pure-NumPy browser build. See the full LQE repository for unavailable modules.


```python
import numpy as np
from lqe import VQE, tfi_hamiltonian
H = tfi_hamiltonian(2, J=1.0, h=0.5)
out = VQE(2, H, num_layers=1, max_iterations=40).solve()
print(f"Toy molecular Hamiltonian energy: {out['energy']:.6f}")
print("Full OpenFermion/PySCF chemistry is intentionally not bundled in LQE Lite.")
```
