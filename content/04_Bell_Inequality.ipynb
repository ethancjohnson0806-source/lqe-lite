# 04 — Bell Inequality (CHSH)

**Concept.** CHSH bounds classical correlations by 2; quantum mechanics can reach $2\sqrt{2}\approx 2.828$.

**Math.** $S = E(a,b)-E(a,b')+E(a',b)+E(a',b')$

```python
from lqe.core.statevector import StatevectorSim
import numpy as np

def corr(ta, tb):
    s = StatevectorSim(2)
    s.apply('H', 0); s.apply('CNOT', 0, 1)
    s.apply('RY', ta, 0); s.apply('RY', tb, 1)
    p = s.probabilities()
    return float(p[0]+p[3]-p[1]-p[2])

a, ap, b, bp = 0.0, np.pi/2, np.pi/4, 3*np.pi/4
S = corr(a,b) - corr(a,bp) + corr(ap,b) + corr(ap,bp)
print(f'S = {S:.4f}  (classical bound 2, quantum 2.828)')
```

**Takeaway.** Nature violates local realism; CHSH is the experimental proof.
